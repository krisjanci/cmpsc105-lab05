## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette="hsv")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: It was zoomed in so it exagurated the differences.


How you changed the plot and why: We fixed it by setting the graph to go from 0 to max instead of starting at 3000 so that the visual difference is representitive.


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")
diamonds["carat_rounded"] = diamonds["carat"].round(1)

sns.lineplot(
    data=diamonds,
    x="carat_rounded",
    y="price",
    marker="o",
    errorbar=None
    )

plt.title("Diamond Price vs. Carat")
plt.show()
```

```python
diamonds.head()
```

Issues with the plot:
This plot has a large amount of overlaping data point and we can not properly see everything that's going on with the relationship between price and carat

How you changed the plot and why:
We changed the scatterplot to a line plot so that there's not that much clustering and we can directly see how the price is moving as you go up in carats.


## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="month", y="passengers", units="year", estimator=None, hue="year")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: All the lines are same color so you cannot tell them apart and see which year is which line.


How you changed the plot and why: We added `hue="year"` to differentiate lines by year.


## Plot 4

```python
tips = sns.load_dataset("tips")

sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker")
plt.title("Total bill vs Tip Size by Cigarete Use")
plt.show()
```
